# Basic Text Processing with NLTK

## Word tokenization

```python
sent = "This is an example sentence for tokenization."
nltk.word_tokenize(sent)
```

## Sentence tokenization

```python
paragraph = "This is the first sentence. This is the second sentence. This is the third sentence. Is this a fourth sentence? Yes, it is!"
list_of_sentences = nltk.sent_tokenize(paragraph)

for sentence in list_of_sentences:
    print(sentence)
```

## Count words

```python
count_text = "Am I a text? I am filled with words! Words are important. Words are powerful, and they can convey meaning. Words can be used to express thoughts, ideas, and emotions."
tokens = nltk.word_tokenize(count_text)
print("Number of words in the text:", len(tokens))
print("Words used in the text:", tokens)

import string

tokens_without_punctuation = [token for token in tokens if token not in string.punctuation]
print("Number of words in the text without punctuation:", len(tokens_without_punctuation))
print("Words used in the text without punctuation:", tokens_without_punctuation)

print("Number of items separated by whitespace:", len(count_text.split()))
```

Alternative solution

```python
tokens_without_punctuation = []

for token in tokens:
    if token not in string.punctuation:
        tokens_without_punctuation.append(token)

print("Number of words in the text without punctuation:", len(tokens_without_punctuation))
print("Words used in the text without punctuation:", tokens_without_punctuation)
```

## Count sentences

```python
sent_tokenized = nltk.sent_tokenize(count_text)
print("Number of sentences in the text:", len(sent_tokenized))
```

## Lowercase

Lowercasing can help when counting word frequencies because it ensures that words with different capitalizations are treated as the same word. For example, "The" and "the" would be counted as two separate words if they were not lowercased, but if they are lowercased, they will be counted as the same word.

```python
lowercase_text = count_text.lower()
print("Lowercased text:", lowercase_text)
```

## Remove stopwords

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
filtered_tokens = [token for token in tokens_without_punctuation if token not in stop_words]
print("Number of words in the text before removing stopwords:", len(tokens_without_punctuation))
print("Number of words in the text after removing stopwords:", len(filtered_tokens))
print("Words used in the text after removing stopwords:", filtered_tokens)
```

## Most Common Words

```python
common_text = "This is a sample text with some common words. The purpose of this text is to demonstrate how to find the most common words in a given text. The text contains several words that are repeated multiple times, such as 'text', 'common', and 'words'. By analyzing the frequency of these words, we can identify which words are the most common in the text."

# Lowercase the text 
common_text = common_text.lower()

# Tokenize the text into words
common_tokens = nltk.word_tokenize(common_text)

# Remove punctuation from the list of tokens
common_tokens = [token for token in common_tokens if token not in string.punctuation]

# Compare the approaches of using Counter and FreqDist
# Using Counter
counter = Counter(common_tokens)

# Using FreqDist
freq_dist = nltk.FreqDist(common_tokens)

# Check the output types
print("Type of counter:", type(counter))
print("Type of freq_dist:", type(freq_dist))

# Print the 10 most common words using both approaches
print("Most common words using Counter:")
print(counter.most_common(10))
print("\nMost common words using FreqDist:")
print(freq_dist.most_common(10))

```

Both approaches yield the same results.

# Stemming and Lemmatization

## Stemming

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words_to_stem = ["connect", "connected", "connecting", "connection"]

for word in words_to_stem:
    stemmed_word = stemmer.stem(word)
    print(f"Original word: {word}, Stemmed word: {stemmed_word}")
```

## Stemming a text

```python
text_to_stem = "The connected devices are connecting to the network, and the connection is stable."

# Tokenize the text
tokens = nltk.word_tokenize(text_to_stem)

# Apply stemming to each token
stemmed_tokens = [stemmer.stem(token) for token in tokens]

# Alternative using a for loop
stemmed_tokens_loop = []
for token in tokens:
    stemmed_tokens_loop.append(stemmer.stem(token))

print(f"Original text: {text_to_stem}")
print(f"Stemmed text (list comprehension): { ' '.join(stemmed_tokens) }")
print(f"Stemmed text (for loop): { ' '.join(stemmed_tokens_loop) }")
```

Returning stems are not always real words. Stemming reduces words to their base form, but they do not always produce valid words. In the example above "device" is stemmed to "devic" or "stable" is stemmed to "stabl". 

## Lemmatization

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words_to_lemmatize = ["running", "ran", "easily", "fairly", "better", "cats", "went", "studies"]

lemmatized_words = [lemmatizer.lemmatize(word) for word in words_to_lemmatize]

print(f"Original words: {words_to_lemmatize}")
print(f"Lemmatized words default: {lemmatized_words}")

# Specifying the part of speech for lemmatization
lemmatized_words_with_pos_v = [lemmatizer.lemmatize(word, pos='v') for word in words_to_lemmatize]
lemmatized_words_with_pos_n = [lemmatizer.lemmatize(word, pos='n') for word in words_to_lemmatize]
lemmatized_words_with_pos_a = [lemmatizer.lemmatize(word, pos='a') for word in words_to_lemmatize]

print(f"Lemmatized words with pos='v': {lemmatized_words_with_pos_v}")
print(f"Lemmatized words with pos='n': {lemmatized_words_with_pos_n}")
print(f"Lemmatized words with pos='a': {lemmatized_words_with_pos_a}")
```

Specifying the part of speech is important in lemmatization because the meaning of a word can change depending on its part of speech. For example, running with the default setting remains running, while using pos='v' returns run.

# Combined Practical Assignment

```python
# read the text from the file
with open("assignment.txt", "r", encoding="utf-8") as file:
    assignment_text = file.read()

# convert the text to lowercase
assignment_text_lower = assignment_text.lower()

# split the text into sentences and tokens
assignment_sentences = nltk.sent_tokenize(assignment_text_lower)
assignment_tokens = nltk.word_tokenize(assignment_text_lower)

# remove punctuation from the list of tokens
assignment_tokens_without_punctuation = [token for token in assignment_tokens if token not in string.punctuation]

# remove stopwords
stop_words = set(stopwords.words('english'))
assignment_tokens_without_stopwords = [token for token in assignment_tokens_without_punctuation if token not in stop_words]

# display the number of sentences, tokens, tokens without punctuation, and tokens without stopwords
print("Number of sentences in the assignment text:", len(assignment_sentences))
print("Number of tokens in the assignment text:", len(assignment_tokens))
print("Number of tokens without punctuation in the assignment text:", len(assignment_tokens_without_punctuation))
print("Number of tokens without stopwords in the assignment text:", len(assignment_tokens_without_stopwords))

# count and show the 10 most common words
counter_assignment = Counter(assignment_tokens_without_stopwords)
print("\n10 most common words in the assignment text:", counter_assignment.most_common(10))

# perform stemming and lemmatization and compare the result with the original
# perform stemming
stemmed_assignment_tokens = [stemmer.stem(token) for token in assignment_tokens_without_stopwords] 

# perform lemmatization
lemmatized_assignment_tokens = [lemmatizer.lemmatize(token) for token in assignment_tokens_without_stopwords]

# compare the results
print("\nOriginal tokens without stopwords:", assignment_tokens_without_stopwords)
print("Stemmed tokens:", stemmed_assignment_tokens)
print("Lemmatized tokens:", lemmatized_assignment_tokens)
```
